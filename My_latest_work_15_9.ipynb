```python
import json
import time
import difflib
import textwrap
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import train_test_split

import torch
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer


BANKING_TRAIN_URL = "https://raw.githubusercontent.com/PolyAI-LDN/task-specific-datasets/master/banking_data/train.csv"
BANKING_TEST_URL = "https://raw.githubusercontent.com/PolyAI-LDN/task-specific-datasets/master/banking_data/test.csv"

MODEL_NAMES = {
    "flan-t5-small": "google/flan-t5-small",
    "flan-t5-base": "google/flan-t5-base",
    "flan-t5-large": "google/flan-t5-large",
}
LLM_MODEL_NAME = MODEL_NAMES["flan-t5-base"]

RANDOM_STATE = 42
TEST_SIZE = 0.25
EVALUATION_SAMPLE_SIZE = 300
CONFIDENCE_THRESHOLD = 0.60
SIMULATED_PARTICIPANTS = 30

WORKFLOW_POLICY = {}
def normalize_columns(df):
    df = df.copy()
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_", regex=False)
        .str.replace("-", "_", regex=False)
    )
    return df


def find_column(df, possible_names):
    for name in possible_names:
        if name in df.columns:
            return name
    raise ValueError(
        "Required column not found. Tried: " + ", ".join(possible_names)
        + "\nAvailable columns: " + ", ".join(df.columns)
    )


def load_and_prepare_data(train_url=BANKING_TRAIN_URL, test_url=BANKING_TEST_URL):
    train_df = pd.read_csv(train_url)
    test_df = pd.read_csv(test_url)
    df = pd.concat([train_df, test_df], ignore_index=True)
    df = normalize_columns(df)

    description_col = find_column(df, ["text", "ticket_description", "description", "query", "message"])
    type_col = find_column(df, ["category", "ticket_type", "type", "intent", "issue_type"])

    df["input_text"] = df[description_col].fillna("").astype(str)
    data = df[["input_text", type_col]].copy()
    data = data.rename(columns={type_col: "ticket_type"})

    data["input_text"] = data["input_text"].fillna("").astype(str)
    data["ticket_type"] = data["ticket_type"].fillna("").astype(str)

    data = data[data["input_text"].str.strip().str.len() > 5]
    data = data[data["ticket_type"].str.strip().str.len() > 0]

    np.random.seed(RANDOM_STATE)
    urgent_keywords = ["urgent", "asap", "immediately", "emergency", "stolen", "fraud", "scam"]

    def assign_priority(text):
        text_lower = text.lower()
        if any(word in text_lower for word in urgent_keywords):
            return "High"
        return str(np.random.choice(["Medium", "Low"]))

    data["ticket_priority"] = data["input_text"].apply(assign_priority)

    class_counts = data["ticket_type"].value_counts()
    valid_classes = class_counts[class_counts >= 2].index
    data = data[data["ticket_type"].isin(valid_classes)].reset_index(drop=True)

    print("========== DATA LOADED ==========")
    print("Records used:", len(data))
    print("Ticket types (BANKING77 intents):", data["ticket_type"].nunique())

    return data


data = load_and_prepare_data()
def build_dynamic_workflow_policy(data):
    ticket_types = sorted(data["ticket_type"].unique())
    priorities = sorted(data["ticket_priority"].unique())

    policy = {
        "ticket_types": ticket_types,
        "priorities": priorities,
        "confidence_threshold": CONFIDENCE_THRESHOLD,
        "routing_rules": {},
        "priority_rules": {},
    }

    for ticket_type in ticket_types:
        route_name = str(ticket_type).strip().lower().replace(" ", "_").replace("/", "_").replace("-", "_")
        policy["routing_rules"][ticket_type] = {
            "workflow_name": f"{route_name}_workflow",
            "owner_team": f"{ticket_type} Support Team",
            "requires_audit_log": True,
        }

    for priority in priorities:
        priority_text = str(priority).lower()
        if any(word in priority_text for word in ["critical", "urgent", "high"]):
            priority_level, sla, escalation = "high", "Immediate review required", "Escalate to senior support team"
        elif any(word in priority_text for word in ["medium", "normal"]):
            priority_level, sla, escalation = "medium", "Standard business queue", "Assign to standard support queue"
        else:
            priority_level, sla, escalation = "low", "Normal support queue", "Assign to normal support queue"

        policy["priority_rules"][priority] = {
            "priority_level": priority_level, "sla": sla, "escalation_action": escalation,
        }

    print("\n========== DYNAMIC WORKFLOW POLICY CREATED ==========")
    print("Ticket workflows:", len(policy["routing_rules"]))
    print("Priority rules:", len(policy["priority_rules"]))

    with open("dynamic_workflow_policy.json", "w", encoding="utf-8") as file:
        json.dump(policy, file, indent=4)

    return policy


WORKFLOW_POLICY = build_dynamic_workflow_policy(data)


from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
import joblib

def train_supervised_baseline(train_df):
    vectorizer = TfidfVectorizer(ngram_range=(1, 2), min_df=2, sublinear_tf=True)
    X_train_vec = vectorizer.fit_transform(train_df["input_text"])

    clf = LogisticRegression(max_iter=2000, C=10)
    clf.fit(X_train_vec, train_df["ticket_type"])

    joblib.dump(vectorizer, "tfidf_vectorizer.joblib")
    joblib.dump(clf, "trained_ticket_classifier.joblib")

    print("========== SUPERVISED BASELINE TRAINED ==========")
    print("Vectorizer vocabulary size:", len(vectorizer.vocabulary_))
    print("Classes learned:", len(clf.classes_))
    print("Saved: tfidf_vectorizer.joblib, trained_ticket_classifier.joblib")

    return vectorizer, clf


def supervised_predict(vectorizer, clf, texts):
    vec = vectorizer.transform(texts)
    proba = clf.predict_proba(vec)
    preds = clf.classes_[proba.argmax(axis=1)]
    confidences = proba.max(axis=1)
    return preds, confidences


def load_llm(model_name):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForSeq2SeqLM.from_pretrained(model_name).to(device)
    model.eval()
    print("\n========== LLM LOADED ==========")
    print("LLM model:", model_name)
    print("Device:", device)
    return {"tokenizer": tokenizer, "model": model, "device": device, "model_name": model_name}


def generate_text(llm, prompt, max_new_tokens=40):
    tokenizer, model, device = llm["tokenizer"], llm["model"], llm["device"]
    inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=512).to(device)
    with torch.no_grad():
        output_ids = model.generate(**inputs, max_new_tokens=max_new_tokens, do_sample=False)
    return tokenizer.decode(output_ids[0], skip_special_tokens=True).strip()


def build_few_shot_examples(train_df, labels, per_class=1, seed=RANDOM_STATE):
    rng = np.random.RandomState(seed)
    examples = []
    for label in labels:
        subset = train_df[train_df["ticket_type"] == label]
        if len(subset) == 0:
            continue
        row = subset.sample(1, random_state=seed).iloc[0]
        examples.append(f"Ticket: {row['input_text']}\nType: {label}")
    rng.shuffle(examples)
    return examples[:8]


def make_classification_prompt(ticket_text, priority, labels, few_shot_examples=None):
    label_text = ", ".join(labels)
    examples_block = ""
    if few_shot_examples:
        examples_block = "Examples:\n" + "\n\n".join(few_shot_examples) + "\n\n"

    return f"""You are an LLM-powered AI agent for business process automation.
Classify the customer support ticket into exactly ONE ticket type from the allowed list.
Reply with ONLY the ticket type text, nothing else.

Allowed ticket types:
{label_text}

{examples_block}Ticket priority: {priority}
Ticket text: {ticket_text}

Ticket type:"""
def normalize_prediction(raw_prediction, labels):
    raw = str(raw_prediction).strip()
    raw_lower = raw.lower()
    for label in labels:
        if raw_lower == label.lower():
            return label
    for label in labels:
        if label.lower() in raw_lower or raw_lower in label.lower():
            return label
    close = difflib.get_close_matches(raw, labels, n=1, cutoff=0.0)
    return close[0] if close else labels[0]


def pseudo_confidence(raw_prediction, selected_label):
    raw = str(raw_prediction).strip().lower()
    label = str(selected_label).strip().lower()
    if raw == label:
        return 0.95
    if label in raw or raw in label:
        return 0.80
    similarity = difflib.SequenceMatcher(None, raw, label).ratio()
    return round(max(0.30, min(0.75, similarity)), 4)


def llm_classify_ticket(llm, ticket_text, priority, labels, few_shot_examples=None):
    prompt = make_classification_prompt(ticket_text, priority, labels, few_shot_examples)
    raw_output = generate_text(llm, prompt, max_new_tokens=20)
    predicted_type = normalize_prediction(raw_output, labels)
    confidence = pseudo_confidence(raw_output, predicted_type)
    return predicted_type, confidence, raw_output


def hybrid_classify_ticket(vectorizer, clf, llm, ticket_text, priority, labels, few_shot_examples=None):
    sup_pred, sup_conf = supervised_predict(vectorizer, clf, [ticket_text])
    predicted_type = sup_pred[0]
    confidence = float(sup_conf[0])

    llm_raw_output = None
    if llm is not None:
        _, _, llm_raw_output = llm_classify_ticket(llm, ticket_text, priority, labels, few_shot_examples)

    return predicted_type, confidence, llm_raw_output
def business_action(predicted_type, priority, confidence, workflow_policy):
    routing_rule = workflow_policy["routing_rules"].get(predicted_type, {
        "workflow_name": "general_support_workflow",
        "owner_team": "General Support Team",
        "requires_audit_log": True,
    })
    priority_rule = workflow_policy["priority_rules"].get(priority, {
        "priority_level": "unknown",
        "sla": "Manual SLA review required",
        "escalation_action": "Send to support coordinator",
    })
    review_action = "Automation allowed" if confidence >= workflow_policy["confidence_threshold"] else "Human review required before automation"
    return (
        f"{review_action}; {priority_rule['escalation_action']}; "
        f"route to {routing_rule['workflow_name']}; owner team: {routing_rule['owner_team']}; "
        f"SLA: {priority_rule['sla']}"
    )


def failure_mode(true_type, predicted_type, confidence):
    if true_type == predicted_type and confidence >= CONFIDENCE_THRESHOLD:
        return "success"
    if true_type == predicted_type and confidence < CONFIDENCE_THRESHOLD:
        return "low_confidence_success"
    if true_type != predicted_type and confidence >= CONFIDENCE_THRESHOLD:
        return "wrong_class_high_confidence"
    return "wrong_class_low_confidence"


def error_recovery_action(mode):
    recovery = {
        "success": "Proceed with automated workflow and log the decision for audit.",
        "low_confidence_success": "Ask one clarifying question, then route after confirmation.",
        "wrong_class_low_confidence": "Stop automation, send to human review, and request missing details.",
        "wrong_class_high_confidence": "Critical failure: block automation, escalate to supervisor, and log for model review.",
    }
    return recovery.get(mode, "Send to human review and log unknown failure.")


def reliability_score(mode):
    scores = {
        "success": 1.00, "low_confidence_success": 0.75,
        "wrong_class_low_confidence": 0.35, "wrong_class_high_confidence": 0.00,
    }
    return scores.get(mode, 0.00)


def trust_label(score):
    if score >= 0.80:
        return "high_trust"
    if score >= 0.50:
        return "medium_trust"
    return "low_trust"

import random

MOCK_ACCOUNT_DB = {
    "default": {"balance": 1523.40, "currency": "GBP", "card_status": "active"}
}

def mock_account_lookup(customer_id="default"):
    account = MOCK_ACCOUNT_DB.get(customer_id, MOCK_ACCOUNT_DB["default"])
    return {"tool": "account_lookup", "customer_id": customer_id, **account}


def mock_card_status_check(customer_id="default"):
    account = MOCK_ACCOUNT_DB.get(customer_id, MOCK_ACCOUNT_DB["default"])
    return {"tool": "card_status_check", "customer_id": customer_id, "status": account["card_status"]}


def mock_transfer_eta():
    return {"tool": "transfer_eta_lookup", "eta_hours": random.choice([1, 2, 24, 48])}


TICKET_TYPE_TO_TOOL = {
    "card": mock_card_status_check,
    "balance": mock_account_lookup,
    "transfer": mock_transfer_eta,
    "top_up": mock_account_lookup,
}


def run_business_tool(predicted_type, customer_id="default"):
    predicted_lower = predicted_type.lower()
    for keyword, tool_fn in TICKET_TYPE_TO_TOOL.items():
        if keyword in predicted_lower:
            try:
                return tool_fn(customer_id) if tool_fn is not mock_transfer_eta else tool_fn()
            except TypeError:
                return tool_fn()
    return {"tool": "none", "note": "no matching business tool for this ticket type"}
from sklearn.metrics import precision_recall_fscore_support

def compute_prf(y_true, y_pred):
    precision_macro, recall_macro, f1_macro, _ = precision_recall_fscore_support(
        y_true, y_pred, average="macro", zero_division=0
    )
    precision_weighted, recall_weighted, f1_weighted, _ = precision_recall_fscore_support(
        y_true, y_pred, average="weighted", zero_division=0
    )
    return {
        "precision_macro": round(precision_macro, 4),
        "recall_macro": round(recall_macro, 4),
        "f1_macro": round(f1_macro, 4),
        "precision_weighted": round(precision_weighted, 4),
        "recall_weighted": round(recall_weighted, 4),
        "f1_weighted": round(f1_weighted, 4),
    }
def evaluate_approach(test_df, labels, predict_fn, workflow_policy, approach_name):
    results = []
    start_time = time.time()

    for index, row in test_df.reset_index(drop=True).iterrows():
        predicted_type, confidence, raw_output = predict_fn(row)
        mode = failure_mode(row["ticket_type"], predicted_type, confidence)
        recovery = error_recovery_action(mode)
        score = reliability_score(mode)

        results.append({
            "input_text": row["input_text"],
            "ticket_priority": row["ticket_priority"],
            "true_ticket_type": row["ticket_type"],
            "raw_output": raw_output,
            "predicted_ticket_type": predicted_type,
            "confidence": confidence,
            "business_action": business_action(predicted_type, row["ticket_priority"], confidence, workflow_policy),
            "failure_mode": mode,
            "error_recovery_action": recovery,
            "reliability_score": score,
            "human_trust_estimate": trust_label(score),
        })

        if (index + 1) % 200 == 0:
            print(f"[{approach_name}] processed {index + 1}/{len(test_df)}")

    avg_latency = (time.time() - start_time) / max(len(test_df), 1)
    result_df = pd.DataFrame(results)

    accuracy = accuracy_score(result_df["true_ticket_type"], result_df["predicted_ticket_type"])
    task_success_rate = (result_df["failure_mode"] == "success").mean()
    avg_reliability = result_df["reliability_score"].mean()
    prf = compute_prf(result_df["true_ticket_type"], result_df["predicted_ticket_type"])

    summary = {
        "approach": approach_name,
        "records_evaluated": int(len(test_df)),
        "accuracy": float(round(accuracy, 4)),
        "task_success_rate": float(round(task_success_rate, 4)),
        "failure_rate": float(round(1 - task_success_rate, 4)),
        "average_reliability_score": float(round(avg_reliability, 4)),
        "average_latency_seconds": float(round(avg_latency, 4)),
        **prf,
    }

    print(f"\n========== {approach_name.upper()} RESULTS ==========")
    for k, v in summary.items():
        print(f"{k}: {v}")
    print("\nFailure mode breakdown:")
    print(result_df["failure_mode"].value_counts())

    return result_df, summary


def run_full_comparison(data, workflow_policy, run_llm=True, llm_model_names=None):
    if llm_model_names is None:
        llm_model_names = list(MODEL_NAMES.values())

    train_df, test_df = train_test_split(
        data, test_size=TEST_SIZE, random_state=RANDOM_STATE, stratify=data["ticket_type"]
    )
    labels = sorted(train_df["ticket_type"].unique())

    if EVALUATION_SAMPLE_SIZE is not None and len(test_df) > EVALUATION_SAMPLE_SIZE:
        test_df_sample = test_df.sample(EVALUATION_SAMPLE_SIZE, random_state=RANDOM_STATE)
    else:
        test_df_sample = test_df

    vectorizer, clf = train_supervised_baseline(train_df)

    def supervised_predict_fn(row):
        preds, confs = supervised_predict(vectorizer, clf, [row["input_text"]])
        return preds[0], float(confs[0]), None

    sup_result_df, sup_summary = evaluate_approach(
        test_df_sample, labels, supervised_predict_fn, workflow_policy, "trained_supervised_baseline"
    )
    sup_summary["llm_model_name"] = None

    all_summaries = [sup_summary]
    all_result_dfs = {"trained_supervised_baseline": sup_result_df}

    if run_llm:
        few_shot = build_few_shot_examples(train_df, labels)

        for llm_model_name in llm_model_names:
            short_name = llm_model_name.split("/")[-1]
            llm = load_llm(llm_model_name)

            def llm_predict_fn(row, llm=llm):
                return llm_classify_ticket(llm, row["input_text"], row["ticket_priority"], labels, few_shot)

            llm_result_df, llm_summary = evaluate_approach(
                test_df_sample, labels, llm_predict_fn, workflow_policy, f"zero_shot_llm_{short_name}"
            )
            llm_summary["llm_model_name"] = llm_model_name
            all_summaries.append(llm_summary)
            all_result_dfs[f"zero_shot_llm_{short_name}"] = llm_result_df

            def hybrid_predict_fn(row, llm=llm):
                pred, conf, raw = hybrid_classify_ticket(
                    vectorizer, clf, llm, row["input_text"], row["ticket_priority"], labels, few_shot
                )
                return pred, conf, raw

            hybrid_result_df, hybrid_summary = evaluate_approach(
                test_df_sample, labels, hybrid_predict_fn, workflow_policy, f"hybrid_trained_plus_llm_{short_name}"
            )
            hybrid_summary["llm_model_name"] = llm_model_name
            all_summaries.append(hybrid_summary)
            all_result_dfs[f"hybrid_trained_plus_llm_{short_name}"] = hybrid_result_df

            del llm
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

    comparison_df = pd.DataFrame(all_summaries)
    comparison_df.to_csv("model_comparison_summary.csv", index=False)
    print("\n\n========== FINAL COMPARISON ==========")
    print(comparison_df)

    return comparison_df, all_result_dfs, vectorizer, clf, labels
def plot_accuracy_comparison(comparison_df):
    plt.figure(figsize=(8, 5))
    bars = plt.bar(comparison_df["approach"], comparison_df["accuracy"], color="#2E86AB")
    plt.title("Accuracy: Trained Baseline vs Zero-Shot LLM vs Hybrid")
    plt.ylabel("Accuracy")
    plt.ylim(0, 1)
    plt.xticks(rotation=15, ha="right")
    for i, acc in enumerate(comparison_df["accuracy"]):
        plt.text(i, acc + 0.02, f"{acc:.2f}", ha="center")
    plt.tight_layout()
    plt.savefig("accuracy_comparison.png", dpi=200)
    plt.show()


def plot_confusion_matrix(result_df, labels, model_tag="trained_supervised_baseline"):
    labels_all = sorted(set(result_df["true_ticket_type"]) | set(result_df["predicted_ticket_type"]))
    cm = confusion_matrix(result_df["true_ticket_type"], result_df["predicted_ticket_type"], labels=labels_all)
    cm_df = pd.DataFrame(cm, index=labels_all, columns=labels_all)
    cm_df.to_csv(f"confusion_matrix_{model_tag}.csv")

    size = min(12, max(6, len(labels_all) * 0.15))
    plt.figure(figsize=(size, size * 0.85))
    plt.imshow(cm_df.values, cmap="Blues")
    plt.title(f"Confusion Matrix -- {model_tag}")
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.colorbar(label="Count")
    plt.tight_layout()
    plt.savefig(f"confusion_matrix_{model_tag}.png", dpi=150)
    plt.show()


def plot_failure_modes(result_df, model_tag="trained_supervised_baseline"):
    counts = result_df["failure_mode"].value_counts()
    plt.figure(figsize=(9, 5))
    bars = plt.barh(counts.index[::-1], counts.values[::-1], color="#2E86AB")
    plt.title(f"Failure Modes -- {model_tag}")
    plt.xlabel("Number of Tickets")
    for bar in bars:
        plt.text(bar.get_width() + 0.5, bar.get_y() + bar.get_height() / 2,
                  int(bar.get_width()), va="center")
    plt.tight_layout()
    plt.savefig(f"failure_modes_{model_tag}.png", dpi=200)
    plt.show()


SIMULATED_DEMO = True


def simulate_human_trust_survey(average_reliability, n_participants=SIMULATED_PARTICIPANTS):
    np.random.seed(RANDOM_STATE)
    rows = []
    for participant_id in range(1, n_participants + 1):
        pre_trust = np.random.randint(3, 6)
        if average_reliability >= 0.80:
            change = np.random.choice([1, 2], p=[0.45, 0.55])
        elif average_reliability >= 0.50:
            change = np.random.choice([-1, 0, 1], p=[0.20, 0.45, 0.35])
        else:
            change = np.random.choice([-2, -1, 0], p=[0.35, 0.45, 0.20])
        post_trust = int(max(1, min(7, pre_trust + change)))
        rows.append({
            "participant_id": participant_id,
            "pre_interaction_trust_1_to_7": int(pre_trust),
            "post_interaction_trust_1_to_7": post_trust,
            "trust_change": post_trust - int(pre_trust),
            "perceived_reliability_condition": trust_label(average_reliability),
            "data_source": "SIMULATED_DEMO",
        })
    return pd.DataFrame(rows)


def make_real_trust_survey_template(n_rows=15):
    template = pd.DataFrame({
        "participant_id": range(1, n_rows + 1),
        "pre_interaction_trust_1_to_7": [""] * n_rows,
        "post_interaction_trust_1_to_7": [""] * n_rows,
        "comments": [""] * n_rows,
    })
    template.to_csv("real_trust_survey_template.csv", index=False)
    print("Template saved: real_trust_survey_template.csv")
    print("Give this to real participants after they try a few sample tickets with the agent.")
    return template


def load_real_trust_survey(path="real_trust_survey_template.csv"):
    df = pd.read_csv(path)
    df["pre_interaction_trust_1_to_7"] = df["pre_interaction_trust_1_to_7"].astype(float)
    df["post_interaction_trust_1_to_7"] = df["post_interaction_trust_1_to_7"].astype(float)
    df["trust_change"] = df["post_interaction_trust_1_to_7"] - df["pre_interaction_trust_1_to_7"]
    df["data_source"] = "REAL_PARTICIPANT"
    return df
def predict_new_ticket(vectorizer, clf, workflow_policy, subject, description, priority,
                        llm=None, labels=None, few_shot=None, customer_id="default"):
    text = (subject + " " + description).strip()
    preds, confs = supervised_predict(vectorizer, clf, [text])
    predicted_type, confidence = preds[0], float(confs[0])

    mode = "success" if confidence >= CONFIDENCE_THRESHOLD else "low_confidence_success"
    recovery = error_recovery_action(mode)
    action = business_action(predicted_type, priority, confidence, workflow_policy)
    tool_result = run_business_tool(predicted_type, customer_id)

    response_text = None
    if llm is not None:
        prompt = f"""You are a professional customer support agent.
Ticket type: {predicted_type}
Priority: {priority}
Ticket text: {text}
Tool data: {tool_result}
Internal action: {recovery}

Write a short, professional response to the customer:"""
        response_text = generate_text(llm, prompt, max_new_tokens=80)

    print("\n========== NEW TICKET PREDICTION ==========")
    print("Text:", text)
    print("Predicted ticket type:", predicted_type, "| confidence:", round(confidence, 3))
    print("Business action:", action)
    print("Business tool called:", tool_result)
    print("Error recovery action:", recovery)
    if response_text:
        print("LLM-generated response:", response_text)

    return {
        "predicted_ticket_type": predicted_type,
        "confidence": confidence,
        "business_action": action,
        "tool_result": tool_result,
        "response_text": response_text,
    }
import random

def perturb_text(text, mode):
    chars = list(text)
    if mode == "typo" and len(chars) > 3:
        i = random.randint(0, len(chars) - 2)
        chars[i], chars[i + 1] = chars[i + 1], chars[i]
        return "".join(chars)
    if mode == "case":
        return text.upper() if random.random() < 0.5 else text.lower()
    if mode == "whitespace":
        return "  " + text.replace(" ", "   ") + "  "
    if mode == "truncate":
        cut = max(3, int(len(text) * 0.6))
        return text[:cut]
    if mode == "noise":
        noise_chars = "!@#$%*&"
        pos = random.randint(0, len(text))
        return text[:pos] + random.choice(noise_chars) + text[pos:]
    return text


def robustness_test(vectorizer, clf, test_df, modes=None, seed=RANDOM_STATE):
    if modes is None:
        modes = ["typo", "case", "whitespace", "truncate", "noise"]
    random.seed(seed)
    base_preds, _ = supervised_predict(vectorizer, clf, test_df["input_text"])
    base_acc = accuracy_score(test_df["ticket_type"], base_preds)
    rows = [{"scenario": "clean", "accuracy": round(base_acc, 4), "accuracy_drop": 0.0}]
    for mode in modes:
        perturbed = test_df["input_text"].apply(lambda t: perturb_text(t, mode))
        preds, _ = supervised_predict(vectorizer, clf, perturbed)
        acc = accuracy_score(test_df["ticket_type"], preds)
        rows.append({"scenario": mode, "accuracy": round(acc, 4), "accuracy_drop": round(base_acc - acc, 4)})
    result = pd.DataFrame(rows)
    result.to_csv("robustness_test_results.csv", index=False)
    print(result)
    return result
BUSINESS_SCENARIOS = {
    "normal_mix": {"High": 0.15, "Medium": 0.45, "Low": 0.40},
    "peak_high_priority": {"High": 0.50, "Medium": 0.30, "Low": 0.20},
    "low_priority_bulk": {"High": 0.05, "Medium": 0.25, "Low": 0.70},
}


def sample_scenario(test_df, distribution, n, seed):
    frames = []
    for priority, frac in distribution.items():
        subset = test_df[test_df["ticket_priority"] == priority]
        take = min(len(subset), max(1, int(n * frac)))
        if take > 0:
            frames.append(subset.sample(take, random_state=seed))
    if not frames:
        return test_df.sample(min(n, len(test_df)), random_state=seed)
    return pd.concat(frames).sample(frac=1, random_state=seed).reset_index(drop=True)


def run_multi_scenario_evaluation(data, vectorizer, clf, workflow_policy, n_per_scenario=150, n_repeats=3):
    train_df, test_df = train_test_split(
        data, test_size=TEST_SIZE, random_state=RANDOM_STATE, stratify=data["ticket_type"]
    )
    rows = []
    for scenario_name, distribution in BUSINESS_SCENARIOS.items():
        for repeat in range(n_repeats):
            seed = RANDOM_STATE + repeat
            scenario_df = sample_scenario(test_df, distribution, n_per_scenario, seed)
            preds, confs = supervised_predict(vectorizer, clf, scenario_df["input_text"])
            acc = accuracy_score(scenario_df["ticket_type"], preds)
            prf = compute_prf(scenario_df["ticket_type"], preds)
            rows.append({
                "scenario": scenario_name,
                "repeat": repeat + 1,
                "n": len(scenario_df),
                "accuracy": round(acc, 4),
                **prf,
            })
    result = pd.DataFrame(rows)
    result.to_csv("multi_scenario_evaluation.csv", index=False)
    print(result)
    return result
def compute_reliability_framework(result_df, robustness_df, multi_scenario_df):
    task_success_rate = (result_df["failure_mode"] == "success").mean()
    robustness_score = 1 - robustness_df[robustness_df["scenario"] != "clean"]["accuracy_drop"].mean()
    scenario_std = multi_scenario_df.groupby("scenario")["accuracy"].std().mean()
    consistency_score = max(0.0, 1 - scenario_std)
    avg_confidence = result_df["confidence"].mean()

    reliability_index = (
        0.4 * task_success_rate
        + 0.25 * robustness_score
        + 0.2 * consistency_score
        + 0.15 * avg_confidence
    )

    framework = {
        "task_success_rate": round(float(task_success_rate), 4),
        "robustness_score": round(float(robustness_score), 4),
        "consistency_score": round(float(consistency_score), 4),
        "average_confidence": round(float(avg_confidence), 4),
        "reliability_index": round(float(reliability_index), 4),
    }

    with open("reliability_framework_summary.json", "w", encoding="utf-8") as f:
        json.dump(framework, f, indent=4)

    print(framework)
    return framework
def run_real_trust_study(vectorizer, clf, sample_tickets, n_participants=5):
    rows = []
    for participant_id in range(1, n_participants + 1):
        pre_trust = int(input(f"Participant {participant_id} - trust in AI agents before test (1-7): "))
        for ticket in sample_tickets:
            preds, confs = supervised_predict(vectorizer, clf, [ticket])
            print("Ticket:", ticket)
            print("Predicted category:", preds[0], "| confidence:", round(float(confs[0]), 3))
        post_trust = int(input(f"Participant {participant_id} - trust in this AI agent after test (1-7): "))
        comment = input("Any comments: ")
        rows.append({
            "participant_id": participant_id,
            "pre_interaction_trust_1_to_7": pre_trust,
            "post_interaction_trust_1_to_7": post_trust,
            "trust_change": post_trust - pre_trust,
            "comments": comment,
            "data_source": "REAL_PARTICIPANT",
        })
    result = pd.DataFrame(rows)
    result.to_csv("human_trust_survey_REAL.csv", index=False)
    print(result)
    return result
RUN_LLM = True

comparison_df, all_result_dfs, vectorizer, clf, labels = run_full_comparison(
    data, WORKFLOW_POLICY, run_llm=RUN_LLM, llm_model_names=list(MODEL_NAMES.values())
)

plot_accuracy_comparison(comparison_df)

for approach_name, rdf in all_result_dfs.items():
    plot_confusion_matrix(rdf, labels, approach_name)
    plot_failure_modes(rdf, approach_name)
    rdf.to_csv(f"evaluation_results_{approach_name}.csv", index=False)

best_row = comparison_df.loc[comparison_df["accuracy"].idxmax()]
if SIMULATED_DEMO:
    trust_df = simulate_human_trust_survey(best_row["average_reliability_score"])
    trust_df.to_csv("human_trust_survey_SIMULATED.csv", index=False)
    print("\nSaved SIMULATED trust data -> human_trust_survey_SIMULATED.csv")
    print("Remember to label this as simulated/demonstration data in your thesis.")
else:
    trust_df = load_real_trust_survey("real_trust_survey_template.csv")
    trust_df.to_csv("human_trust_survey_REAL.csv", index=False)

make_real_trust_survey_template()

run_business_tool("card_arrival")
run_business_tool("balance_not_updated_after_bank_transfer")

predict_new_ticket(
    vectorizer, clf, WORKFLOW_POLICY,
    subject="", description="Payment failed during checkout, money was deducted but the order was not confirmed.",
    priority="High",
)
predict_new_ticket(
    vectorizer, clf, WORKFLOW_POLICY,
    subject="", description="How long does it usually take for a bank transfer to show up in my balance?",
    priority="Medium",
)

train_df_split, test_df_split = train_test_split(
    data, test_size=TEST_SIZE, random_state=RANDOM_STATE, stratify=data["ticket_type"]
)
test_df_split = test_df_split.sample(min(len(test_df_split), EVALUATION_SAMPLE_SIZE), random_state=RANDOM_STATE)

robustness_df = robustness_test(vectorizer, clf, test_df_split)
multi_scenario_df = run_multi_scenario_evaluation(data, vectorizer, clf, WORKFLOW_POLICY)
reliability_framework = compute_reliability_framework(
    all_result_dfs["trained_supervised_baseline"], robustness_df, multi_scenario_df
)

RUN_REAL_TRUST_STUDY = False
if RUN_REAL_TRUST_STUDY:
    sample_tickets = test_df_split["input_text"].sample(3, random_state=RANDOM_STATE).tolist()
    run_real_trust_study(vectorizer, clf, sample_tickets, n_participants=5)

print("\n\n========== ALL FILES SAVED ==========")
print("model_comparison_summary.csv, dynamic_workflow_policy.json,")
print("trained_ticket_classifier.joblib, tfidf_vectorizer.joblib,")
print("evaluation_results_*.csv, confusion_matrix_*.csv/.png,")
print("accuracy_comparison.png, failure_modes_*.png,")
print("real_trust_survey_template.csv, human_trust_survey_*.csv,")
print("robustness_test_results.csv, multi_scenario_evaluation.csv,")
print("reliability_framework_summary.json")
```
