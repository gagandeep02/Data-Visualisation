```python
"""
Building and Evaluating LLM-Powered AI Agents for Business Process Automation:
Failure Modes, Reliability, and Human Trust

Dataset : BANKING77 (Casanueva et al., ACL 2020 NLP4ConvAI) - CC-BY-4.0
Source  : https://github.com/PolyAI-LDN/task-specific-datasets
Model   : google/flan-t5-small (single model; see MODEL_NAME to change)
"""

import json
import time
import gc
from pathlib import Path

import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer
from transformers.modeling_outputs import BaseModelOutput


# ============================================================================
# CONFIGURATION
# ============================================================================

BANKING77_BASE = (
    "https://raw.githubusercontent.com/PolyAI-LDN/"
    "task-specific-datasets/master/banking_data/"
)

# Models are loaded and freed one at a time. All three see the IDENTICAL
# evaluation sample (drawn once, outside the model loop) so results can be
# compared with paired statistics.
MODEL_NAMES = [
    "google/flan-t5-small",   #  80M
    "google/flan-t5-base",    # 250M
    "google/flan-t5-large",   # 780M
]

# flan-t5-large in fp32 is ~3.1GB. Use fp16 on GPU; leave fp32 on CPU
# (fp16 on CPU is slow and numerically unstable).
USE_FP16_ON_GPU = True

RANDOM_STATE = 42
EVALUATION_SAMPLE_SIZE = 300     # None = use the full 3,080-row test split
LABEL_BATCH_SIZE = 32            # labels scored per forward pass

# The agent abstains (defers to a human) below this confidence.
# Section 8 sweeps this across its whole range; 0.50 is only the
# operating point used for the headline tables.
ABSTAIN_THRESHOLD = 0.50

SIMULATED_PARTICIPANTS = 30

OUTPUT_DIR = Path(".")


# ============================================================================
# 1. DATA LOADING
# ============================================================================

def load_banking77():
    """Load BANKING77 from its official GitHub source.

    Uses the authors' published train/test split rather than a new random
    split, so results stay comparable with published numbers.
    """
    train = pd.read_csv(BANKING77_BASE + "train.csv")
    test = pd.read_csv(BANKING77_BASE + "test.csv")

    for df in (train, test):
        df.rename(columns={"text": "input_text", "category": "intent"}, inplace=True)
        df["input_text"] = df["input_text"].fillna("").astype(str).str.strip()
        df["intent"] = df["intent"].fillna("").astype(str).str.strip()

    train = train[train["input_text"].str.len() > 3].reset_index(drop=True)
    test = test[test["input_text"].str.len() > 3].reset_index(drop=True)

    print("========== DATA LOADED ==========")
    print("Dataset       : BANKING77 (PolyAI, CC-BY-4.0)")
    print("Train records :", len(train))
    print("Test records  :", len(test))
    print("Intent classes:", train["intent"].nunique())
    print("Median words  :", int(train["input_text"].str.split().str.len().median()))

    return train, test


def supervised_ceiling_test(train, test):
    """Confirm the label is learnable from the text before trusting any LLM result.

    This is the check that disqualified the original Kaggle dataset, where a
    supervised model could not beat the majority-class baseline. Report this
    table in the methodology chapter as a dataset-validation step.
    """
    vec = TfidfVectorizer(max_features=20000, ngram_range=(1, 2), sublinear_tf=True)
    a = vec.fit_transform(train["input_text"])
    b = vec.transform(test["input_text"])

    clf = LogisticRegression(max_iter=2000, C=5).fit(a, train["intent"])
    ceiling = accuracy_score(test["intent"], clf.predict(b))
    majority = test["intent"].value_counts(normalize=True).max()
    chance = 1.0 / train["intent"].nunique()

    print("\n========== DATASET VALIDATION (SUPERVISED CEILING) ==========")
    print("TF-IDF + LogReg ceiling :", round(ceiling, 4))
    print("Majority-class baseline :", round(majority, 4))
    print("Chance baseline         :", round(chance, 4))

    if ceiling < majority + 0.05:
        print("WARNING: label appears unlearnable from text. Do not proceed.")
    else:
        print("PASS: the label carries real signal.")

    return {
        "supervised_ceiling": float(round(ceiling, 4)),
        "majority_baseline": float(round(majority, 4)),
        "chance_baseline": float(round(chance, 4)),
    }


# ============================================================================
# 2. BUSINESS PROCESS POLICY
# ============================================================================
# BANKING77 has no priority column. In real ITSM practice priority is not an
# input field but a value assigned by policy from the ticket's category and
# business impact. The mapping below is authored explicitly, which makes the
# workflow policy an actual business process model rather than string
# templating over label names.
#
# REVISE THESE ASSIGNMENTS. Defending your own reasoning about which banking
# intents are Critical is far stronger than defending a mapping you inherited.

INTENT_ROUTE = {
    # --- Fraud and security -------------------------------------------------
    "compromised_card": "fraud_and_security",
    "lost_or_stolen_card": "fraud_and_security",
    "lost_or_stolen_phone": "fraud_and_security",
    "card_swallowed": "fraud_and_security",
    "cash_withdrawal_not_recognised": "fraud_and_security",
    "card_payment_not_recognised": "fraud_and_security",
    "direct_debit_payment_not_recognised": "fraud_and_security",
    "transaction_charged_twice": "fraud_and_security",
    "pin_blocked": "fraud_and_security",
    "passcode_forgotten": "fraud_and_security",

    # --- Card operations ----------------------------------------------------
    "activate_my_card": "card_operations",
    "card_about_to_expire": "card_operations",
    "card_acceptance": "card_operations",
    "card_arrival": "card_operations",
    "card_delivery_estimate": "card_operations",
    "card_linking": "card_operations",
    "card_not_working": "card_operations",
    "contactless_not_working": "card_operations",
    "get_physical_card": "card_operations",
    "getting_spare_card": "card_operations",
    "getting_virtual_card": "card_operations",
    "get_disposable_virtual_card": "card_operations",
    "order_physical_card": "card_operations",
    "virtual_card_not_working": "card_operations",
    "change_pin": "card_operations",
    "disposable_card_limits": "card_operations",
    "visa_or_mastercard": "card_operations",
    "supported_cards_and_currencies": "card_operations",
    "apple_pay_or_google_pay": "card_operations",

    # --- Payments and transfers --------------------------------------------
    "declined_card_payment": "payments_and_transfers",
    "declined_cash_withdrawal": "payments_and_transfers",
    "declined_transfer": "payments_and_transfers",
    "failed_transfer": "payments_and_transfers",
    "cancel_transfer": "payments_and_transfers",
    "beneficiary_not_allowed": "payments_and_transfers",
    "pending_card_payment": "payments_and_transfers",
    "pending_cash_withdrawal": "payments_and_transfers",
    "pending_transfer": "payments_and_transfers",
    "transfer_into_account": "payments_and_transfers",
    "transfer_not_received_by_recipient": "payments_and_transfers",
    "transfer_timing": "payments_and_transfers",
    "receiving_money": "payments_and_transfers",
    "reverted_card_payment?": "payments_and_transfers",
    "balance_not_updated_after_bank_transfer": "payments_and_transfers",
    "balance_not_updated_after_cheque_or_cash_deposit": "payments_and_transfers",
    "wrong_amount_of_cash_received": "payments_and_transfers",

    # --- Fees and charges ---------------------------------------------------
    "card_payment_fee_charged": "fees_and_charges",
    "cash_withdrawal_charge": "fees_and_charges",
    "exchange_charge": "fees_and_charges",
    "extra_charge_on_statement": "fees_and_charges",
    "transfer_fee_charged": "fees_and_charges",
    "top_up_by_bank_transfer_charge": "fees_and_charges",
    "top_up_by_card_charge": "fees_and_charges",
    "card_payment_wrong_exchange_rate": "fees_and_charges",
    "wrong_exchange_rate_for_cash_withdrawal": "fees_and_charges",
    "exchange_rate": "fees_and_charges",

    # --- Top-up and currency ------------------------------------------------
    "automatic_top_up": "topup_and_currency",
    "pending_top_up": "topup_and_currency",
    "top_up_failed": "topup_and_currency",
    "top_up_limits": "topup_and_currency",
    "top_up_reverted": "topup_and_currency",
    "topping_up_by_card": "topup_and_currency",
    "top_up_by_cash_or_cheque": "topup_and_currency",
    "verify_top_up": "topup_and_currency",
    "exchange_via_app": "topup_and_currency",
    "fiat_currency_support": "topup_and_currency",
    "atm_support": "topup_and_currency",
    "country_support": "topup_and_currency",

    # --- Refunds ------------------------------------------------------------
    "Refund_not_showing_up": "refunds",
    "request_refund": "refunds",

    # --- Account administration --------------------------------------------
    "age_limit": "account_administration",
    "edit_personal_details": "account_administration",
    "terminate_account": "account_administration",
    "unable_to_verify_identity": "account_administration",
    "verify_my_identity": "account_administration",
    "verify_source_of_funds": "account_administration",
    "why_verify_identity": "account_administration",
}

INTENT_PRIORITY = {
    # Critical - active financial loss or security compromise
    "compromised_card": "Critical",
    "lost_or_stolen_card": "Critical",
    "lost_or_stolen_phone": "Critical",
    "cash_withdrawal_not_recognised": "Critical",
    "card_payment_not_recognised": "Critical",
    "direct_debit_payment_not_recognised": "Critical",

    # High - customer blocked from funds, or money missing
    "card_swallowed": "High",
    "pin_blocked": "High",
    "passcode_forgotten": "High",
    "transaction_charged_twice": "High",
    "wrong_amount_of_cash_received": "High",
    "transfer_not_received_by_recipient": "High",
    "Refund_not_showing_up": "High",
    "declined_card_payment": "High",
    "declined_cash_withdrawal": "High",
    "declined_transfer": "High",
    "failed_transfer": "High",
    "top_up_failed": "High",
    "card_not_working": "High",
    "virtual_card_not_working": "High",
    "contactless_not_working": "High",
    "balance_not_updated_after_bank_transfer": "High",
    "balance_not_updated_after_cheque_or_cash_deposit": "High",
    "unable_to_verify_identity": "High",

    # Medium - action required, no immediate loss
    "request_refund": "Medium",
    "cancel_transfer": "Medium",
    "beneficiary_not_allowed": "Medium",
    "pending_card_payment": "Medium",
    "pending_cash_withdrawal": "Medium",
    "pending_transfer": "Medium",
    "pending_top_up": "Medium",
    "reverted_card_payment?": "Medium",
    "top_up_reverted": "Medium",
    "activate_my_card": "Medium",
    "change_pin": "Medium",
    "card_about_to_expire": "Medium",
    "edit_personal_details": "Medium",
    "terminate_account": "Medium",
    "verify_my_identity": "Medium",
    "verify_source_of_funds": "Medium",
    "verify_top_up": "Medium",
    "card_payment_fee_charged": "Medium",
    "cash_withdrawal_charge": "Medium",
    "exchange_charge": "Medium",
    "extra_charge_on_statement": "Medium",
    "transfer_fee_charged": "Medium",
    "top_up_by_bank_transfer_charge": "Medium",
    "top_up_by_card_charge": "Medium",
    "card_payment_wrong_exchange_rate": "Medium",
    "wrong_exchange_rate_for_cash_withdrawal": "Medium",
    "transfer_into_account": "Medium",
    "receiving_money": "Medium",

    # Low - informational
    "card_arrival": "Low",
    "card_delivery_estimate": "Low",
    "card_acceptance": "Low",
    "card_linking": "Low",
    "get_physical_card": "Low",
    "getting_spare_card": "Low",
    "getting_virtual_card": "Low",
    "get_disposable_virtual_card": "Low",
    "order_physical_card": "Low",
    "disposable_card_limits": "Low",
    "visa_or_mastercard": "Low",
    "supported_cards_and_currencies": "Low",
    "apple_pay_or_google_pay": "Low",
    "exchange_rate": "Low",
    "exchange_via_app": "Low",
    "fiat_currency_support": "Low",
    "atm_support": "Low",
    "country_support": "Low",
    "transfer_timing": "Low",
    "top_up_limits": "Low",
    "topping_up_by_card": "Low",
    "top_up_by_cash_or_cheque": "Low",
    "automatic_top_up": "Low",
    "age_limit": "Low",
    "why_verify_identity": "Low",
}

PRIORITY_RANK = {"Low": 0, "Medium": 1, "High": 2, "Critical": 3}

PRIORITY_SLA = {
    "Critical": ("Immediate action, 15 minute response",
                 "Escalate to fraud and security desk"),
    "High":     ("Same-day response, 4 hour target",
                 "Escalate to senior support analyst"),
    "Medium":   ("Standard business queue, 1 working day",
                 "Assign to standard support queue"),
    "Low":      ("Normal queue, 3 working days",
                 "Assign to self-service or normal queue"),
}

ROUTE_TEAM = {
    "fraud_and_security": "Fraud and Security Team",
    "card_operations": "Card Operations Team",
    "payments_and_transfers": "Payments and Transfers Team",
    "fees_and_charges": "Fees and Billing Team",
    "topup_and_currency": "Top-up and Currency Team",
    "refunds": "Refunds Team",
    "account_administration": "Account Administration Team",
}


def build_workflow_policy(labels):
    """Assemble the routing policy and verify every intent is covered."""
    missing_route = [x for x in labels if x not in INTENT_ROUTE]
    missing_prio = [x for x in labels if x not in INTENT_PRIORITY]
    if missing_route or missing_prio:
        raise ValueError(
            f"Unmapped intents.\nNo route: {missing_route}\nNo priority: {missing_prio}"
        )

    policy = {
        "intents": list(labels),
        "abstain_threshold": ABSTAIN_THRESHOLD,
        "routing_rules": {},
        "priority_rules": {},
    }

    for intent in labels:
        route = INTENT_ROUTE[intent]
        priority = INTENT_PRIORITY[intent]
        sla, escalation = PRIORITY_SLA[priority]

        policy["routing_rules"][intent] = {
            "workflow_name": f"{route}_workflow",
            "route": route,
            "owner_team": ROUTE_TEAM[route],
            "requires_audit_log": True,
        }
        policy["priority_rules"][intent] = {
            "priority": priority,
            "priority_rank": PRIORITY_RANK[priority],
            "sla": sla,
            "escalation_action": escalation,
        }

    print("\n========== WORKFLOW POLICY BUILT ==========")
    print("Intents mapped     :", len(policy["routing_rules"]))
    print("Operational routes :", len(set(INTENT_ROUTE.values())))
    print("Priority levels    :", sorted(PRIORITY_RANK, key=PRIORITY_RANK.get))
    counts = pd.Series([INTENT_PRIORITY[i] for i in labels]).value_counts()
    print("Intents per priority:\n" + counts.to_string())

    with open(OUTPUT_DIR / "workflow_policy.json", "w", encoding="utf-8") as f:
        json.dump(policy, f, indent=4)

    return policy


# ============================================================================
# 3. LLM LOADING AND LABEL SCORING
# ============================================================================

def load_llm(model_name):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    dtype = (torch.float16
             if (USE_FP16_ON_GPU and device.type == "cuda")
             else torch.float32)

    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForSeq2SeqLM.from_pretrained(
        model_name, torch_dtype=dtype).to(device)
    model.eval()

    print("\n========== LLM LOADED ==========")
    print("Model :", model_name)
    print("Device:", device, "| dtype:", dtype)
    print("Params:", f"{sum(p.numel() for p in model.parameters()) / 1e6:.1f}M")

    return {"tokenizer": tokenizer, "model": model,
            "device": device, "name": model_name}


def free_llm(llm):
    """Release GPU memory before loading the next model."""
    del llm["model"]
    llm.clear()
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()


def make_classification_prompt(ticket_text):
    """Prompt for intent classification.

    The 77 label names are deliberately NOT listed here. Scoring is done by
    likelihood over the fixed label set (see score_labels), so the model is
    constrained to valid labels by construction. Listing all 77 names would
    consume 600-900 tokens and overflow the 512-token encoder limit, silently
    truncating the customer query.
    """
    return (
        "Classify the intent of this online banking customer query.\n\n"
        f"Query: {ticket_text}\n\n"
        "Intent:"
    )


def prepare_label_cache(llm, labels):
    """Tokenize the label set once and reuse it for every ticket."""
    tokenizer = llm["tokenizer"]
    readable = [lbl.replace("_", " ").replace("?", "").strip() for lbl in labels]
    enc = tokenizer(readable, return_tensors="pt", padding=True)
    return {
        "labels": list(labels),
        "input_ids": enc.input_ids.to(llm["device"]),
        "pad_id": tokenizer.pad_token_id,
    }


def score_labels(llm, ticket_text, label_cache):
    """Return (predicted_intent, confidence, full probability vector).

    Confidence is a genuine posterior: the length-normalized sequence
    log-likelihood of each candidate label under the model, softmaxed across
    the 77 candidates. This replaces the earlier string-similarity heuristic,
    which was circular (it measured how closely the generated string matched
    the label that had just been selected as its closest match).
    """
    tokenizer, model, device = llm["tokenizer"], llm["model"], llm["device"]
    label_ids = label_cache["input_ids"]
    pad_id = label_cache["pad_id"]
    n_labels = label_ids.shape[0]

    prompt = make_classification_prompt(ticket_text)
    enc = tokenizer(prompt, return_tensors="pt",
                    truncation=True, max_length=512).to(device)

    with torch.no_grad():
        encoder_out = model.get_encoder()(**enc)
        hidden = encoder_out.last_hidden_state

        seq_logprobs = []
        for start in range(0, n_labels, LABEL_BATCH_SIZE):
            chunk = label_ids[start:start + LABEL_BATCH_SIZE]
            k = chunk.shape[0]

            expanded = BaseModelOutput(
                last_hidden_state=hidden.expand(k, -1, -1).contiguous()
            )
            attn = enc.attention_mask.expand(k, -1).contiguous()
            masked = chunk.masked_fill(chunk == pad_id, -100)

            out = model(encoder_outputs=expanded,
                        attention_mask=attn, labels=masked)

            logp = torch.log_softmax(out.logits.float(), dim=-1)
            tok_lp = logp.gather(2, chunk.unsqueeze(-1)).squeeze(-1)
            mask = (chunk != pad_id).float()
            seq_logprobs.append((tok_lp * mask).sum(1) / mask.sum(1))

        scores = torch.cat(seq_logprobs)
        probs = torch.softmax(scores, dim=0)

    best = int(probs.argmax())
    return label_cache["labels"][best], float(probs[best]), probs.cpu().numpy()


# ============================================================================
# 4. FAILURE MODES AND SEVERITY-WEIGHTED RELIABILITY
# ============================================================================

def failure_mode(true_intent, pred_intent, confidence, policy):
    """Four-way taxonomy built around the abstention decision.

    The key asymmetry: an error the agent catches and defers is a success of
    the safety mechanism, whereas an error it acts on confidently is the
    failure that matters for automation safety.
    """
    correct = (true_intent == pred_intent)
    automated = confidence >= policy["abstain_threshold"]

    if correct and automated:
        return "correct_automated"
    if correct and not automated:
        return "correct_deferred"          # unnecessary human effort
    if not correct and not automated:
        return "incorrect_deferred"        # error caught by abstention
    return "incorrect_automated"           # dangerous: acted on a wrong label


def severity_weighted_reliability(true_intent, pred_intent, mode, policy):
    """Reliability in [0, 1], weighted by the business cost of the error.

    Replaces fixed hand-picked constants. Severity is derived from the
    workflow policy: an error that keeps the ticket in the same team at the
    same priority is nearly harmless, while one that routes a Critical case
    into a Low-priority queue is the worst outcome in the system.
    """
    if mode == "correct_automated":
        return 1.00
    if mode == "correct_deferred":
        return 0.70          # correct, but consumed human review time
    if mode == "incorrect_deferred":
        return 0.60          # wrong, but the agent knew and handed off

    # incorrect_automated: cost depends on where the ticket actually landed
    true_route = policy["routing_rules"][true_intent]["route"]
    pred_route = policy["routing_rules"][pred_intent]["route"]
    true_rank = policy["priority_rules"][true_intent]["priority_rank"]
    pred_rank = policy["priority_rules"][pred_intent]["priority_rank"]

    score = 0.50 if true_route == pred_route else 0.30

    drop = true_rank - pred_rank
    if drop > 0:                       # under-prioritized: the dangerous case
        score -= 0.15 * drop
    elif drop < 0:                     # over-prioritized: wastes capacity only
        score -= 0.05 * abs(drop)

    return float(max(0.0, round(score, 4)))


def error_recovery_action(mode, true_intent, pred_intent, policy):
    if mode == "correct_automated":
        return "Proceed with automated workflow and log the decision for audit."
    if mode == "correct_deferred":
        return "Ask one clarifying question, then route after confirmation."
    if mode == "incorrect_deferred":
        return "Stop automation, send to human review, request missing details."

    true_rank = policy["priority_rules"][true_intent]["priority_rank"]
    pred_rank = policy["priority_rules"][pred_intent]["priority_rank"]
    if true_rank - pred_rank >= 2:
        return ("Critical failure: high-priority case routed to a low-priority "
                "queue. Block automation, escalate to supervisor, log for review.")
    return ("Misrouted with high confidence. Block automation, escalate to "
            "supervisor, and log for model review.")


def trust_label(score):
    if score >= 0.80:
        return "high_trust"
    if score >= 0.50:
        return "medium_trust"
    return "low_trust"


def business_action(pred_intent, confidence, policy):
    routing = policy["routing_rules"][pred_intent]
    priority = policy["priority_rules"][pred_intent]

    gate = ("Automation allowed" if confidence >= policy["abstain_threshold"]
            else "Human review required before automation")

    return (f"{gate}; {priority['escalation_action']}; "
            f"route to {routing['workflow_name']}; "
            f"owner team: {routing['owner_team']}; "
            f"priority: {priority['priority']}; SLA: {priority['sla']}")


# ============================================================================
# 5. CALIBRATION AND UNCERTAINTY METRICS
# ============================================================================

def wilson_interval(successes, n, z=1.96):
    """95% CI for a proportion. Report this beside every accuracy figure."""
    if n == 0:
        return (0.0, 0.0)
    p = successes / n
    denom = 1 + z ** 2 / n
    centre = (p + z ** 2 / (2 * n)) / denom
    margin = z * np.sqrt(p * (1 - p) / n + z ** 2 / (4 * n ** 2)) / denom
    return (float(max(0.0, centre - margin)), float(min(1.0, centre + margin)))


def expected_calibration_error(confidences, correctness, n_bins=10):
    """ECE: mean gap between stated confidence and observed accuracy.

    Low accuracy with low ECE means an agent that knows it is unsure, which
    is safe. Low accuracy with high ECE means confident errors, which is not.
    """
    confidences = np.asarray(confidences, dtype=float)
    correctness = np.asarray(correctness, dtype=float)
    edges = np.linspace(0.0, 1.0, n_bins + 1)

    ece = 0.0
    rows = []
    for i in range(n_bins):
        lo, hi = edges[i], edges[i + 1]
        sel = (confidences > lo) & (confidences <= hi) if i > 0 else \
              (confidences >= lo) & (confidences <= hi)
        if sel.sum() == 0:
            rows.append({"bin_low": lo, "bin_high": hi, "n": 0,
                         "mean_confidence": np.nan, "accuracy": np.nan})
            continue
        acc = correctness[sel].mean()
        conf = confidences[sel].mean()
        ece += (sel.sum() / len(confidences)) * abs(acc - conf)
        rows.append({"bin_low": lo, "bin_high": hi, "n": int(sel.sum()),
                     "mean_confidence": float(conf), "accuracy": float(acc)})

    return float(ece), pd.DataFrame(rows)


def risk_coverage_curve(confidences, correctness, n_points=50):
    """Accuracy on automated tickets as a function of automation coverage.

    This is the selective-prediction view of the abstention threshold and
    answers the operational question directly: at a given automation rate,
    how accurate is the agent on what it chose to automate?
    """
    confidences = np.asarray(confidences, dtype=float)
    correctness = np.asarray(correctness, dtype=float)
    order = np.argsort(-confidences)
    sorted_correct = correctness[order]
    sorted_conf = confidences[order]

    n = len(confidences)
    rows = []
    for k in np.unique(np.linspace(1, n, n_points).astype(int)):
        rows.append({
            "threshold": float(sorted_conf[k - 1]),
            "coverage": float(k / n),
            "accuracy_on_automated": float(sorted_correct[:k].mean()),
            "n_automated": int(k),
        })
    return pd.DataFrame(rows)


# ============================================================================
# 6. CUSTOMER RESPONSE GENERATION
# ============================================================================

def generate_customer_response(llm, ticket_text, pred_intent, policy,
                               max_new_tokens=80):
    """Generate a customer-facing reply.

    NOTE: this output is not evaluated anywhere in the pipeline. Either add a
    human rating on a small sample, or scope generation out of your claims.
    Do not report it as a validated capability.
    """
    routing = policy["routing_rules"][pred_intent]
    priority = policy["priority_rules"][pred_intent]

    prompt = (
        "Write a short professional reply to this online banking customer.\n\n"
        f"Customer message: {ticket_text}\n"
        f"Identified issue: {pred_intent.replace('_', ' ')}\n"
        f"Handled by: {routing['owner_team']}\n"
        f"Response target: {priority['sla']}\n\n"
        "Reply:"
    )

    tokenizer, model, device = llm["tokenizer"], llm["model"], llm["device"]
    enc = tokenizer(prompt, return_tensors="pt",
                    truncation=True, max_length=512).to(device)
    with torch.no_grad():
        out = model.generate(**enc, max_new_tokens=max_new_tokens, do_sample=False)
    return tokenizer.decode(out[0], skip_special_tokens=True).strip()


# ============================================================================
# 7. SIMULATED TRUST QUESTIONNAIRE
# ============================================================================

def simulate_human_trust_survey(average_reliability):
    """SYNTHETIC DATA - NOT A HUMAN STUDY.

    Responses are drawn from hardcoded probabilities conditioned on the
    agent's reliability. This demonstrates the measurement instrument; it
    does not measure trust. No inferential statistic computed on this frame
    is meaningful, and every table or figure derived from it must be
    labelled synthetic. To make a real trust claim, replace this with data
    collected using a validated scale (e.g. Jian et al. 2000, Trust in
    Automated Systems).
    """
    print("\n" + "!" * 70)
    print("WARNING: generating SYNTHETIC trust responses. Not a human study.")
    print("!" * 70)

    rng = np.random.default_rng(RANDOM_STATE)
    rows = []
    for pid in range(1, SIMULATED_PARTICIPANTS + 1):
        pre = int(rng.integers(3, 6))
        if average_reliability >= 0.80:
            change = int(rng.choice([1, 2], p=[0.45, 0.55]))
        elif average_reliability >= 0.50:
            change = int(rng.choice([-1, 0, 1], p=[0.20, 0.45, 0.35]))
        else:
            change = int(rng.choice([-2, -1, 0], p=[0.35, 0.45, 0.20]))
        post = int(max(1, min(7, pre + change)))

        rows.append({
            "participant_id": pid,
            "data_source": "SYNTHETIC",
            "pre_interaction_trust_1_to_7": pre,
            "post_interaction_trust_1_to_7": post,
            "trust_change": post - pre,
            "perceived_reliability_condition": trust_label(average_reliability),
        })
    return pd.DataFrame(rows)


# ============================================================================
# 8. EVALUATION
# ============================================================================

def draw_evaluation_sample(test_df):
    """Draw the evaluation set ONCE, outside the model loop.

    Every model must see identical items in identical order, otherwise the
    paired statistics below are invalid and the comparison is worthless.
    """
    eval_df = test_df
    if EVALUATION_SAMPLE_SIZE is not None and len(test_df) > EVALUATION_SAMPLE_SIZE:
        eval_df = test_df.sample(EVALUATION_SAMPLE_SIZE,
                                 random_state=RANDOM_STATE)
    eval_df = eval_df.reset_index(drop=True)

    eval_df.to_csv(OUTPUT_DIR / "shared_evaluation_sample.csv", index=False)
    print("\n========== SHARED EVALUATION SAMPLE ==========")
    print("Tickets:", len(eval_df), "(identical for every model)")
    print("Saved to shared_evaluation_sample.csv for auditability")
    return eval_df


def evaluate_agent(llm, eval_df, policy, labels):
    label_cache = prepare_label_cache(llm, labels)

    print("\n========== EVALUATING AGENT ==========")
    print("Model            :", llm["name"])
    print("Evaluated tickets:", len(eval_df))
    print("Candidate intents:", len(labels))
    print("Forward passes   :", len(eval_df) * len(labels))

    results = []
    prob_matrix = np.zeros((len(eval_df), len(labels)), dtype=np.float32)
    start = time.time()

    for i, row in eval_df.iterrows():
        pred, conf, probs = score_labels(llm, row["input_text"], label_cache)
        prob_matrix[i] = probs

        mode = failure_mode(row["intent"], pred, conf, policy)
        score = severity_weighted_reliability(row["intent"], pred, mode, policy)

        results.append({
            "input_text": row["input_text"],
            "true_intent": row["intent"],
            "predicted_intent": pred,
            "confidence": round(conf, 6),
            "true_route": policy["routing_rules"][row["intent"]]["route"],
            "predicted_route": policy["routing_rules"][pred]["route"],
            "true_priority": policy["priority_rules"][row["intent"]]["priority"],
            "predicted_priority": policy["priority_rules"][pred]["priority"],
            "correct": int(row["intent"] == pred),
            "failure_mode": mode,
            "reliability_score": score,
            "human_trust_estimate": trust_label(score),
            "business_action": business_action(pred, conf, policy),
            "error_recovery_action": error_recovery_action(
                mode, row["intent"], pred, policy),
        })

        if (i + 1) % 25 == 0:
            elapsed = time.time() - start
            rate = (i + 1) / elapsed
            print(f"  {i + 1}/{len(eval_df)} tickets "
                  f"({rate:.1f}/s, ~{(len(eval_df) - i - 1) / rate:.0f}s left)")

    latency = (time.time() - start) / len(eval_df)
    result_df = pd.DataFrame(results)

    # Cache raw probabilities so every downstream metric can be recomputed
    # without re-running inference.
    np.savez_compressed(
        OUTPUT_DIR / f"probs_{llm['name'].split('/')[-1]}.npz",
        probs=prob_matrix,
        labels=np.array(labels),
        true_intent=eval_df["intent"].values,
    )

    return result_df, latency


def report_metrics(result_df, latency, model_name, ceiling_info):
    n = len(result_df)
    correct = int(result_df["correct"].sum())
    accuracy = correct / n
    lo, hi = wilson_interval(correct, n)

    ece, calib_df = expected_calibration_error(
        result_df["confidence"], result_df["correct"])
    rc_df = risk_coverage_curve(result_df["confidence"], result_df["correct"])

    route_acc = (result_df["true_route"] == result_df["predicted_route"]).mean()
    prio_acc = (result_df["true_priority"] == result_df["predicted_priority"]).mean()

    counts = result_df["failure_mode"].value_counts()
    automated = counts.get("correct_automated", 0) + counts.get("incorrect_automated", 0)
    dangerous = counts.get("incorrect_automated", 0)

    print("\n========== AGENT PERFORMANCE ==========")
    print("Model                    :", model_name)
    print("Evaluated tickets        :", n)
    print(f"Intent accuracy          : {accuracy:.4f}  (95% CI {lo:.4f}-{hi:.4f})")
    print(f"Supervised ceiling       : {ceiling_info['supervised_ceiling']:.4f}")
    print(f"Chance baseline          : {ceiling_info['chance_baseline']:.4f}")
    print(f"Route accuracy           : {route_acc:.4f}")
    print(f"Priority accuracy        : {prio_acc:.4f}")
    print(f"Expected calibration err : {ece:.4f}")
    print(f"Mean reliability (wtd)   : {result_df['reliability_score'].mean():.4f}")
    print(f"Automation coverage      : {automated / n:.4f}")
    print(f"Dangerous failure rate   : {dangerous / n:.4f}  (confident + wrong)")
    print(f"Latency per ticket       : {latency:.3f} s")

    print("\n========== FAILURE MODES ==========")
    print(counts.to_string())

    print("\n========== ESTIMATED TRUST BANDS ==========")
    print(result_df["human_trust_estimate"].value_counts().to_string())

    print("\n========== ROUTE-LEVEL CLASSIFICATION REPORT ==========")
    print(classification_report(result_df["true_route"],
                                result_df["predicted_route"], zero_division=0))

    summary = {
        "dataset": "BANKING77",
        "model": model_name,
        "evaluated_tickets": int(n),
        "intent_accuracy": float(round(accuracy, 4)),
        "accuracy_ci_95": [round(lo, 4), round(hi, 4)],
        "route_accuracy": float(round(route_acc, 4)),
        "priority_accuracy": float(round(prio_acc, 4)),
        "expected_calibration_error": float(round(ece, 4)),
        "mean_weighted_reliability": float(round(result_df["reliability_score"].mean(), 4)),
        "automation_coverage": float(round(automated / n, 4)),
        "dangerous_failure_rate": float(round(dangerous / n, 4)),
        "latency_seconds_per_ticket": float(round(latency, 4)),
        "abstain_threshold": ABSTAIN_THRESHOLD,
        **ceiling_info,
    }
    return summary, calib_df, rc_df


# ============================================================================
# 9. VISUALISATION
# ============================================================================

def create_graphs(result_df, calib_df, rc_df, model_tag):
    plt.style.use("default")

    # Failure modes
    counts = result_df["failure_mode"].value_counts()
    plt.figure(figsize=(10, 5))
    bars = plt.barh(counts.index[::-1], counts.values[::-1], color="#2E86AB")
    plt.title("Agent failure modes")
    plt.xlabel("Number of tickets")
    for b in bars:
        plt.text(b.get_width() + 0.5, b.get_y() + b.get_height() / 2,
                 int(b.get_width()), va="center")
    plt.grid(axis="x", linestyle="--", alpha=0.35)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / f"failure_modes_{model_tag}.png", dpi=200)
    plt.show()

    # Reliability diagram (calibration)
    valid = calib_df.dropna(subset=["accuracy"])
    plt.figure(figsize=(6, 6))
    plt.plot([0, 1], [0, 1], "k--", label="Perfect calibration")
    plt.plot(valid["mean_confidence"], valid["accuracy"],
             "o-", color="#E76F51", label="Observed")
    plt.title("Reliability diagram")
    plt.xlabel("Mean predicted confidence")
    plt.ylabel("Observed accuracy")
    plt.legend()
    plt.grid(linestyle="--", alpha=0.35)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / f"calibration_{model_tag}.png", dpi=200)
    plt.show()

    # Risk-coverage curve
    plt.figure(figsize=(7, 5))
    plt.plot(rc_df["coverage"], rc_df["accuracy_on_automated"],
             color="#2A9D8F", linewidth=2)
    plt.title("Risk-coverage: accuracy vs automation rate")
    plt.xlabel("Coverage (fraction automated)")
    plt.ylabel("Accuracy on automated tickets")
    plt.grid(linestyle="--", alpha=0.35)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / f"risk_coverage_{model_tag}.png", dpi=200)
    plt.show()

    # Weighted reliability distribution
    plt.figure(figsize=(7, 5))
    plt.hist(result_df["reliability_score"], bins=20,
             color="#264653", edgecolor="black")
    plt.title("Severity-weighted reliability distribution")
    plt.xlabel("Reliability score")
    plt.ylabel("Number of tickets")
    plt.grid(axis="y", linestyle="--", alpha=0.35)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / f"reliability_{model_tag}.png", dpi=200)
    plt.show()

    # Route-level confusion matrix (7x7 - the 77x77 version is unreadable)
    routes = sorted(set(result_df["true_route"]) | set(result_df["predicted_route"]))
    cm = confusion_matrix(result_df["true_route"],
                          result_df["predicted_route"], labels=routes)
    cm_df = pd.DataFrame(cm, index=routes, columns=routes)

    plt.figure(figsize=(9, 8))
    plt.imshow(cm, cmap="Blues")
    plt.title("Route-level confusion matrix")
    plt.xlabel("Predicted route")
    plt.ylabel("True route")
    short = [r.replace("_", "\n") for r in routes]
    plt.xticks(range(len(routes)), short, fontsize=8)
    plt.yticks(range(len(routes)), short, fontsize=8)
    plt.colorbar(label="Count")
    for i in range(len(routes)):
        for j in range(len(routes)):
            if cm[i, j] > 0:
                plt.text(j, i, cm[i, j], ha="center", va="center", fontsize=9)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / f"confusion_routes_{model_tag}.png", dpi=200)
    plt.show()

    return cm_df


# ============================================================================
# 10. SINGLE TICKET INFERENCE
# ============================================================================

def predict_new_ticket(llm, query, policy, labels, label_cache=None):
    if label_cache is None:
        label_cache = prepare_label_cache(llm, labels)

    pred, conf, probs = score_labels(llm, query, label_cache)
    order = np.argsort(-probs)[:3]

    automated = conf >= policy["abstain_threshold"]
    decision = ("automation_eligible" if automated else "needs_clarification")

    print("\n========== NEW TICKET ==========")
    print("Query           :", query)
    print("Predicted intent:", pred)
    print("Confidence      :", round(conf, 4))
    print("Top 3 intents   :",
          [(labels[i], round(float(probs[i]), 4)) for i in order])
    print("Assigned priority:", policy["priority_rules"][pred]["priority"])
    print("Decision        :", decision)
    print("Business action :", business_action(pred, conf, policy))
    if automated:
        print("Draft reply     :",
              generate_customer_response(llm, query, pred, policy))
    else:
        print("Draft reply     : withheld - confidence below abstention threshold")


# ============================================================================
# 11. CROSS-MODEL COMPARISON
# ============================================================================

def mcnemar_test(correct_a, correct_b):
    """Paired test for two classifiers on the SAME items.

    Only the disagreements carry information: b = A right / B wrong,
    c = A wrong / B right. Because all models see identical tickets, this is
    far more sensitive than comparing independent confidence intervals, which
    would overlap at n=300 and tell you nothing.

    Uses the exact binomial test when b + c is small, and the chi-square
    approximation with continuity correction otherwise.
    """
    from scipy import stats

    a = np.asarray(correct_a, dtype=bool)
    b_arr = np.asarray(correct_b, dtype=bool)

    b = int(np.sum(a & ~b_arr))
    c = int(np.sum(~a & b_arr))

    if b + c == 0:
        return {"b": b, "c": c, "statistic": 0.0, "p_value": 1.0,
                "method": "no disagreements"}

    if b + c < 25:
        p = float(stats.binomtest(b, b + c, 0.5).pvalue)
        return {"b": b, "c": c, "statistic": float(min(b, c)),
                "p_value": p, "method": "exact binomial"}

    stat = (abs(b - c) - 1) ** 2 / (b + c)
    p = float(stats.chi2.sf(stat, df=1))
    return {"b": b, "c": c, "statistic": float(stat),
            "p_value": p, "method": "chi-square (continuity corrected)"}


def holm_bonferroni(p_values):
    """Adjust p-values for multiple comparisons.

    Three models means three pairwise tests, so an unadjusted 0.05 threshold
    inflates the false-positive rate. Holm is uniformly more powerful than
    plain Bonferroni and makes no independence assumption.
    """
    n = len(p_values)
    order = np.argsort(p_values)
    adjusted = np.empty(n, dtype=float)

    running = 0.0
    for rank, idx in enumerate(order):
        value = (n - rank) * p_values[idx]
        running = max(running, value)
        adjusted[idx] = min(1.0, running)
    return adjusted.tolist()


def compare_models(all_results, all_summaries):
    """Build the headline comparison table plus paired significance tests."""
    names = list(all_results.keys())

    rows = []
    for name in names:
        s = all_summaries[name]
        rows.append({
            "model": name,
            "accuracy": s["intent_accuracy"],
            "ci_low": s["accuracy_ci_95"][0],
            "ci_high": s["accuracy_ci_95"][1],
            "route_accuracy": s["route_accuracy"],
            "priority_accuracy": s["priority_accuracy"],
            "ece": s["expected_calibration_error"],
            "weighted_reliability": s["mean_weighted_reliability"],
            "automation_coverage": s["automation_coverage"],
            "dangerous_failure_rate": s["dangerous_failure_rate"],
            "latency_s": s["latency_seconds_per_ticket"],
        })
    comparison_df = pd.DataFrame(rows)

    print("\n" + "=" * 78)
    print("CROSS-MODEL COMPARISON")
    print("=" * 78)
    print(comparison_df.to_string(index=False))

    # Pairwise McNemar across adjacent and extreme pairs
    pairs, tests = [], []
    for i in range(len(names)):
        for j in range(i + 1, len(names)):
            a, b = names[i], names[j]
            res = mcnemar_test(all_results[a]["correct"].values,
                               all_results[b]["correct"].values)
            pairs.append((a, b))
            tests.append(res)

    adjusted = holm_bonferroni([t["p_value"] for t in tests])

    stat_rows = []
    for (a, b), t, p_adj in zip(pairs, tests, adjusted):
        stat_rows.append({
            "model_a": a,
            "model_b": b,
            "a_right_b_wrong": t["b"],
            "a_wrong_b_right": t["c"],
            "p_value": round(t["p_value"], 6),
            "p_holm_adjusted": round(p_adj, 6),
            "significant_at_0.05": p_adj < 0.05,
            "method": t["method"],
        })
    stats_df = pd.DataFrame(stat_rows)

    print("\n---------- PAIRED McNEMAR TESTS (Holm-adjusted) ----------")
    print(stats_df.to_string(index=False))
    print("\nOnly disagreements are counted. A non-significant result means")
    print("the accuracy gap is within what chance would produce at this n.")

    return comparison_df, stats_df


def plot_model_comparison(all_results, all_calib, all_rc, comparison_df):
    """Overlay the three models on the plots that carry the scaling story."""
    colors = ["#2E86AB", "#E76F51", "#2A9D8F", "#8E7DBE"]
    names = list(all_results.keys())

    # Risk-coverage: the headline figure
    plt.figure(figsize=(8, 5.5))
    for k, name in enumerate(names):
        rc = all_rc[name]
        plt.plot(rc["coverage"], rc["accuracy_on_automated"],
                 linewidth=2, color=colors[k % len(colors)],
                 label=name.split("/")[-1])
    plt.title("Risk-coverage by model size")
    plt.xlabel("Coverage (fraction of tickets automated)")
    plt.ylabel("Accuracy on automated tickets")
    plt.legend()
    plt.grid(linestyle="--", alpha=0.35)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "comparison_risk_coverage.png", dpi=200)
    plt.show()

    # Calibration: does scale make the agent better at knowing it is unsure?
    plt.figure(figsize=(6.5, 6.5))
    plt.plot([0, 1], [0, 1], "k--", label="Perfect calibration")
    for k, name in enumerate(names):
        valid = all_calib[name].dropna(subset=["accuracy"])
        plt.plot(valid["mean_confidence"], valid["accuracy"], "o-",
                 color=colors[k % len(colors)], label=name.split("/")[-1])
    plt.title("Calibration by model size")
    plt.xlabel("Mean predicted confidence")
    plt.ylabel("Observed accuracy")
    plt.legend()
    plt.grid(linestyle="--", alpha=0.35)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "comparison_calibration.png", dpi=200)
    plt.show()

    # Accuracy with confidence intervals
    plt.figure(figsize=(7, 5))
    x = np.arange(len(comparison_df))
    heights = comparison_df["accuracy"].values
    err = np.vstack([heights - comparison_df["ci_low"].values,
                     comparison_df["ci_high"].values - heights])
    plt.bar(x, heights, yerr=err, capsize=6,
            color=[colors[i % len(colors)] for i in range(len(x))])
    plt.xticks(x, [n.split("/")[-1] for n in comparison_df["model"]])
    plt.title("Intent accuracy with 95% confidence intervals")
    plt.ylabel("Accuracy")
    plt.grid(axis="y", linestyle="--", alpha=0.35)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "comparison_accuracy.png", dpi=200)
    plt.show()

    # Dangerous failures: does scale reduce confident errors faster than
    # it reduces errors overall? This is the safety-relevant question.
    plt.figure(figsize=(7, 5))
    width = 0.35
    plt.bar(x - width / 2, 1 - comparison_df["accuracy"], width,
            label="Overall error rate", color="#adb5bd")
    plt.bar(x + width / 2, comparison_df["dangerous_failure_rate"], width,
            label="Confident errors (automated)", color="#C1121F")
    plt.xticks(x, [n.split("/")[-1] for n in comparison_df["model"]])
    plt.title("Total errors vs errors the agent acted on")
    plt.ylabel("Fraction of tickets")
    plt.legend()
    plt.grid(axis="y", linestyle="--", alpha=0.35)
    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "comparison_dangerous_failures.png", dpi=200)
    plt.show()


# ============================================================================
# 12. MAIN
# ============================================================================

def main():
    train, test = load_banking77()
    ceiling_info = supervised_ceiling_test(train, test)

    labels = sorted(train["intent"].unique())
    policy = build_workflow_policy(labels)

    # Drawn ONCE. Every model sees these exact tickets in this exact order.
    eval_df = draw_evaluation_sample(test)

    all_results, all_summaries, all_calib, all_rc = {}, {}, {}, {}

    for model_name in MODEL_NAMES:
        model_tag = model_name.split("/")[-1]
        print("\n" + "#" * 78)
        print("# MODEL:", model_name)
        print("#" * 78)

        llm = load_llm(model_name)
        result_df, latency = evaluate_agent(llm, eval_df, policy, labels)
        summary, calib_df, rc_df = report_metrics(
            result_df, latency, model_name, ceiling_info)

        cm_df = create_graphs(result_df, calib_df, rc_df, model_tag)

        result_df.to_csv(OUTPUT_DIR / f"evaluation_results_{model_tag}.csv",
                         index=False)
        calib_df.to_csv(OUTPUT_DIR / f"calibration_bins_{model_tag}.csv",
                        index=False)
        rc_df.to_csv(OUTPUT_DIR / f"risk_coverage_{model_tag}.csv", index=False)
        cm_df.to_csv(OUTPUT_DIR / f"confusion_routes_{model_tag}.csv")
        with open(OUTPUT_DIR / f"summary_{model_tag}.json", "w",
                  encoding="utf-8") as f:
            json.dump(summary, f, indent=4)

        all_results[model_name] = result_df
        all_summaries[model_name] = summary
        all_calib[model_name] = calib_df
        all_rc[model_name] = rc_df

        # Demonstration inference before the model is released from memory.
        label_cache = prepare_label_cache(llm, labels)
        predict_new_ticket(
            llm, "My card was stolen last night and I think someone used it.",
            policy, labels, label_cache)
        predict_new_ticket(
            llm, "How long does a new card usually take to arrive?",
            policy, labels, label_cache)

        free_llm(llm)

    # ---- cross-model analysis --------------------------------------------
    comparison_df, stats_df = compare_models(all_results, all_summaries)
    plot_model_comparison(all_results, all_calib, all_rc, comparison_df)

    comparison_df.to_csv(OUTPUT_DIR / "model_comparison.csv", index=False)
    stats_df.to_csv(OUTPUT_DIR / "mcnemar_tests.csv", index=False)

    # Trust survey is driven by the best model's reliability.
    best = comparison_df.loc[comparison_df["weighted_reliability"].idxmax(), "model"]
    survey_df = simulate_human_trust_survey(
        all_summaries[best]["mean_weighted_reliability"])
    survey_df.to_csv(OUTPUT_DIR / "simulated_trust_survey_SYNTHETIC.csv",
                     index=False)
    print("\nTrust survey generated from best model:", best)

    print("\n========== FILES SAVED ==========")
    for f in sorted(OUTPUT_DIR.glob("*")):
        if f.suffix in {".csv", ".json", ".png", ".npz"}:
            print(" ", f.name)


if __name__ == "__main__":
    main()
```
